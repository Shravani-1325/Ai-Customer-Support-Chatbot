# AI Chatbot for Customer Support 💬🎙️

# 1. Business Understnding

**Industry** : E-commerece / Samall Business Customer Support

**Problem Statement**: Small Businesses cannot afford 24/7 human support teams, yet customers expect instant responses to repetitive queries like order tracking, review, return policies, and billing.

**Bussiness Impact of the Problem:**
- Higher amount of responses -> customer wait time and Fustration -> Lost sales
- Human Agents wasting time on repetitive FAQs
- No support outside working hours

>**Solutions :**

A lightweight AI chatbot that auto-handles the top 7 query categories, freeing human agents for complex issues only.


## 1.1 Objective & Goal

**Objective** : Creating Rule Based + Ml AI Chatbot MVP that classifies customer intent and returns context - appropriate automed responses via a REST API

**Goal :** End-to-end development of a Python-based intelligent chatbot, featuring a custom intent-classification pipeline and REST API deployment for real-time interaction.
